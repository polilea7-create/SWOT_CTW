```python
def plot_datasets(datasets, variable, vminmax, title, extent=None):
    cb_args = dict(
        add_colorbar=True,
        cbar_kwargs={"shrink": 0.3}
    )
    
    plot_kwargs = dict(
        x="longitude",
        y="latitude",
        cmap="Spectral_r",
        vmin=vminmax[0],
        vmax=vminmax[1],
    )

    fig, ax = plt.subplots(figsize=(12, 8), subplot_kw=dict(projection=ccrs.PlateCarree()))
    if extent: ax.set_extent(extent)
    
    for ds in datasets:
        ds[variable].plot.pcolormesh(
            ax=ax,
            **plot_kwargs,
            **cb_args)
        cb_args=dict(add_colorbar=False)
        
    ax.set_title(title)
    ax.coastlines()
    gls = ax.gridlines(draw_labels=True)
    gls.top_labels=False
    gls.right_labels=False

    return ax
```

```python

# 1. Votre fonction de spectre (inchangée)
def yo_spec(x, opt=1, nsmooth=5):
    x = np.array(x)
    x = x[np.isfinite(x)]
    if len(x) < 20: return None, None, None, None # Sécurité
    if opt >= 0: x = x - np.mean(x)
    if opt == 1: x = detrend(x)
    n = len(x)
    fftx = np.fft.fft(x)
    ps_raw = np.abs(fftx[:n // 2]) ** 2
    ps_smooth = uniform_filter1d(ps_raw, size=nsmooth)
    freq = np.fft.fftfreq(n, d=1)[:n // 2]
    return ps_smooth, freq

# 2. Chargement des données concaténées
ds = xr.open_dataset('/Users/ABC/Desktop/swot_exports/swot_pass_1.nc') # Adaptez le nom

# 3. Préparation du binning en latitude (16°S à 30°S)
lat_min, lat_max = -30, -16
bin_size = 0.5 # Taille des tranches de latitude
bins = np.arange(lat_min, lat_max + bin_size, bin_size)

# Initialisation des listes pour le stockage
all_specs = []
lat_centers = []
freqs = None # On stockera la fréquence commune

# 4. Boucle sur les tranches de latitude
for i in range(len(bins) - 1):
    lat_b = bins[i]
    lat_e = bins[i+1]
    
    # Sélection des données dans la tranche
    subset = ds.where((ds.latitude >= lat_b) & (ds.latitude < lat_e), drop=True)
    
    if subset.sizes['num_obs'] > 50: # Seuil minimum de points
        signal = subset['ssha_unfiltered'].values
        
        P, f = yo_spec(signal, opt=1, nsmooth=5)
        
        if P is not None:
            all_specs.append(P * f)
            lat_centers.append((lat_b + lat_e) / 2)
            if freqs is None: freqs = f

# 5. Conversion pour le plot
all_specs = np.array(all_specs)
lat_centers = np.array(lat_centers)

# 6. Visualisation
fig, ax = plt.subplots(figsize=(10, 6))

pcm = ax.pcolormesh(freqs, lat_centers, all_specs, 
                    shading='auto', cmap='Spectral_r')

ax.set_xscale('log')
ax.set_ylabel('Latitude [°S]')
ax.set_xlabel('Fréquence [cycle/point]') # Note: diviser par la résolution spatiale pour avoir cycle/km
ax.set_title('Spectre Latitudinal - SSHA Unfiltered (SWOT)')
plt.colorbar(pcm, ax=ax, label='Spectre x fréquence')
plt.grid(True, which="both", ls='--', lw=0.5)

plt.show()
```
